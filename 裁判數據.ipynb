```python
data = pd.read_excel(excel_file_path, sheet_name=sheet_name)
```

```python
col = ['umpireInChief','bHand','side','inning','coordX','coordY','pType','call','result']
```

```python
df = data[col]
```

```python
name_mapping = {
    'FF':'直球', 
    'CH':'變速球', 
    'FC':'卡特球', 
    'CU':'曲球', 
    'FO':'指叉球',
    'SL':'滑球', 
    'SI':'伸卡球',
    'KN':'彈指曲球', 
    'EP':'小便球'
}

# 新增一列"球員"，根据"Pitcher"列的值匹配中文名
df['球種'] = df['pType'].map(name_mapping)
```

```python
df
```

```python
summary = df.groupby(['umpireInChief', 'result']).size().unstack(fill_value=0)
summary['Total'] =summary.iloc[:, 1:].sum(axis=1)
summary
```

```python
summary.info()
```

```python
summary['PA']=summary['Total']-summary['IGNORE']-summary['ID']
summary['BB%']=summary['uBB']/summary['PA']*100
summary['K%']=summary['SO']/summary['PA']*100
```
