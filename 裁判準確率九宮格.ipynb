```python
# 1. 定義好球帶邊界
STRIKE_ZONE_LEFT = -50
STRIKE_ZONE_RIGHT = 50
STRIKE_ZONE_BOTTOM = -50
STRIKE_ZONE_TOP = 50

def in_strike_zone(x, y):
    # 使用 <= 與 >= 已經完整包含了「碰到線也算」的邏輯
    return (STRIKE_ZONE_LEFT <= x <= STRIKE_ZONE_RIGHT) and \
           (STRIKE_ZONE_BOTTOM <= y <= STRIKE_ZONE_TOP)

# 2. 判斷是否為規則上的好球 (Rulebook Strike)
df["rulebook_strike"] = df.apply(lambda row: in_strike_zone(row["coordX"], row["coordY"]), axis=1)
```

```python
# 定義 Zone 劃分函數
def assign_zone(x, y):
    # 判斷是否在定義的好球帶邊界內
    if STRIKE_ZONE_LEFT <= x <= STRIKE_ZONE_RIGHT and STRIKE_ZONE_BOTTOM <= y <= STRIKE_ZONE_TOP:
        
        # 定義切分點 (將 100 寬度/高度分為三等份)
        # 水平切分點: -16.67, 16.67
        # 垂直切分點: -16.67, 16.67
        
        # 判斷垂直層級 (y 軸)
        if y > 16.6667:     # 上層
            return 7 if x < -16.6667 else (8 if x <= 16.6667 else 9)
        elif y > -16.6667:  # 中層
            return 4 if x < -16.6667 else (5 if x <= 16.6667 else 6)
        else:               # 下層
            return 1 if x < -16.6667 else (2 if x <= 16.6667 else 3)
    
    return np.nan

# 新增 Zone 欄位
df['Zone'] = df.apply(lambda row: assign_zone(row['coordX'], row['coordY']), axis=1)
# 顯示結果
df
```

```python
import os
import matplotlib.pyplot as plt
import matplotlib
import numpy as np
import pandas as pd

# 設定 Matplotlib 使用中文字體
matplotlib.rcParams['font.sans-serif'] = ['Microsoft JhengHei']
matplotlib.rcParams['axes.unicode_minus'] = False 

# Zone 映射 (Matplotlib Table index 0,0 是左上)
# 7,8,9 是上層; 4,5,6 是中層; 1,2,3 是下層
zone_mapping = {
    7: (0, 0), 8: (1, 0), 9: (2, 0),
    4: (0, 1), 5: (1, 1), 6: (2, 1),
    1: (0, 2), 2: (1, 2), 3: (2, 2)
}

output_folder = "主審判準_投打組合分類"
if not os.path.exists(output_folder):
    os.makedirs(output_folder)

# --- 資料預處理與聚合 (解決 KeyError 的關鍵) ---
# 先過濾掉 NaN，避免 groupby 遺失資料
df_clean = df.dropna(subset=['Zone', 'umpireInChief', 'pHand', 'bHand', 'is_correct']).copy()
df_clean['Zone'] = df_clean['Zone'].astype(int)

# 重要：在 groupby 時，必須同時包含 'pHand' 與 'bHand'
# 這樣產出的 accuracy_by_zone 才會包含這兩個欄位，後面的 mask 才不會報錯
accuracy_by_zone = df_clean.groupby(['umpireInChief', 'pHand', 'bHand', 'Zone']).agg(
    accuracy=('is_correct', lambda x: x.mean() * 100)
).reset_index()

# ---------------------------------------------

umpires = accuracy_by_zone["umpireInChief"].unique()
p_hands = ['R', 'L']
b_hands = ['R', 'L']

for umpire in umpires:
    for p_hand in p_hands:
        for b_hand in b_hands:
            
            # 定義檔名與路徑
            filename = f"{umpire}_{p_hand}P_{b_hand}B_Accuracy(投手視角).png"
            filepath = os.path.join(output_folder, filename)
            
            # 檢查檔案是否已存在，避免重複執行
            if os.path.exists(filepath):
                continue

            # 根據主審、投手、打者進行篩選
            mask = (accuracy_by_zone["umpireInChief"] == umpire) & \
                   (accuracy_by_zone["pHand"] == p_hand) & \
                   (accuracy_by_zone["bHand"] == b_hand)
            
            filtered_data = accuracy_by_zone[mask]
            
            # 如果該組合完全沒資料，則跳過不產圖
            if filtered_data.empty:
                continue

            # 初始化 3x3 九宮格
            text_data = np.full((3, 3), "-", dtype=object)

            # 填充數據
            for _, row in filtered_data.iterrows():
                z = row["Zone"]
                if z in zone_mapping:
                    x, y = zone_mapping[z]
                    acc = row["accuracy"]
                    text_data[y, x] = f"{acc:.1f}%" if pd.notna(acc) else "-"

            # 繪圖
            fig, ax = plt.subplots(figsize=(5, 5))
            ax.axis('off')

            title_str = f"主審：{umpire}\n({p_hand}P vs {b_hand}B) 判準率(投手視角)"
            ax.set_title(title_str, fontsize=14, fontweight='bold', pad=20)

            # 建立表格
            table = plt.table(
                cellText=text_data,
                cellLoc='center',
                loc='center'
            )

            table.auto_set_font_size(False)
            table.set_fontsize(16)
            table.scale(1, 4)

            # 儲存
            plt.savefig(filepath, dpi=150, bbox_inches="tight")
            plt.close(fig)

print(f"處理完成！圖片已儲存於「{output_folder}」資料夾。")
```
